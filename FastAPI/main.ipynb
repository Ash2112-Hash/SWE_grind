```python
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel

app = FastAPI() # creates the app client object


# extending BaseModel for our item data objects
# in this case, the items must be provided as json payloads in reqs
class Item(BaseModel):
    text: str
    is_done: bool = False


items = []


# this is a default root GET req
@app.get("/")
def root():
    return "Hello" # must be a valid return type


## creating routes

# standard get for the items
@app.post("/items")  #if item is not mentioned in path, its treated as a query param, so you will need to specify ?item=xxx when calling the endpoint
def add_items(item: Item):
    items.append(item)
    return items


@app.get("/items/{item_id}", response_model=Item) ## here its a path param, so you specify it as a /items/0 for ex
def get_specifc_item(item_id: int):
    try:
        item = items[item_id]
        return item
    except:
        raise HTTPException(status_code = 404, detail = "item not found")



@app.get("/items", response_model=list[Item]) ## here its a path param, so you specify it as a /items/0 for ex
def get_items(limit: int = 10):  #enforcing as default value if not given, otherwise give as ?limit=12
    return items[0:limit]





```
